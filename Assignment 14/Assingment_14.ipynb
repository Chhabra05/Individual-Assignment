```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

DATASET-1

```python
df = pd.read_csv(r"C:\Users\huzai\OneDrive\Documents\Desktop\IBA Assignments\ICA\New folder\pakistani_ecommerce_dataset.csv")
```

```python
df
```

```python
#Q1: Which is the most popular product category? 
df["product_category"].value_counts().idxmax()
```

```python
#Q2: What is the average order value?
df["order_value"].mean().round(2)
```

```python
#Q3: Which city has the highest number of orders?
df["city"].value_counts().idxmax()
```

```python
#Q4: What is the most used payment method? 
df["payment_method"].value_counts().idxmax()
```

```python
#Q5: Draw a bar chart showing average order value by city. 

avg_by_city = df.groupby("city")["order_value"].mean()

avg_by_city.plot(kind="bar", color="red")
plt.title("Average Order Value by City")
plt.xlabel("City")
plt.ylabel("Average Order Value (PKR)")
plt.show()
```

```python
mf = pd.read_csv(r"C:\Users\huzai\OneDrive\Documents\Desktop\IBA Assignments\ICA\New folder\student_lifestyle_dataset.csv")
```

```python
mf
```

```python
#Q1: What is the average number of study hours? 
mf["study_hours"].mean().round(2)
```

```python
#Q2: Who sleeps more on average – males or females? 
mf.groupby("gender")["sleep_hours"].mean().round(2)
```

```python
##Q3: Is there a relationship between screen time and sleep hours? 

mf["screen_time"].corr(mf["sleep_hours"])      
sns.scatterplot(data=mf, x="screen_time", y="sleep_hours")   
plt.show()
```

```python
#Q4: Draw a histogram of daily transport cost. 
plt.hist(mf["transport_cost"], bins=10, edgecolor="black")
plt.title("Daily Transport Cost")
plt.xlabel("Cost (PKR)")
plt.ylabel("Number of Students")
plt.show()
```
