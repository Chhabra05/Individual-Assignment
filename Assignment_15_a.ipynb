```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv(r"C:\Users\huzai\OneDrive\Documents\Desktop\IBA Assignments\ICA\New folder\ride_hailing_dataset.csv")
```

```python
df
```

```python
#Q1: What is the average fare by service type?
df.groupby("service_type")["fare"].mean().round(2)
```

```python
#Q2: Which city has the most trips?
df["city"].value_counts().idxmax()
```

```python
#Q3: What is the average distance per trip?
df["distance_km"].mean().round(2)
```

```python
#Q4: Compare trip ratings by gender.

avg_rating = df.groupby("gender")["trip_rating"].mean().round(2)
print(avg_rating)

avg_rating.plot(kind="bar", color=["steelblue", "orange"])
plt.xlabel("Gender")
plt.ylabel("Trip Rating")
plt.title("Average Trip Rating")
plt.show()
```

```python
#Q5: Make a bar chart of service type vs. average fare

avg_fare = df.groupby("service_type")["fare"].mean().round(2)
print(avg_fare)

avg_fare.plot(kind="bar", color=["coolgreen","lightsabergreen","mustardyellow","grey"])
plt.xlabel("Service Type")
plt.ylabel("Fare")
plt.title("Average")
plt.show()
```
