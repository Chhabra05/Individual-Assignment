```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv(r"C:\Users\huzai\OneDrive\Documents\Desktop\IBA Assignments\ICA\New folder\student_tea_snacks _1_.csv")
```

```python
df
```

```python
#Q1: What is the most preferred tea type?
df.groupby("tea_type")["prefers_tea"].idxmax()
```

```python
#Q2: What is the average number of cups per gender?
df.groupby("gender")["cups_per_day"].mean().round(2)
```

```python
#Q3: Make a bar chart of snack choices.

snack = df["snack_choice"].value_counts()
print(snack)

snack.plot(kind="bar", color=["brown","crimson","gold"])
plt.xlabel("Snacks")
plt.ylabel("Count")
plt.title("Snack Choices")
plt.show()
```

```python
#Q4: Compare tea preference by age group.

pref = df.groupby("age_group")["prefers_tea"].value_counts()
print(pref)

pref.plot(kind="bar", color=["orange","red","blue"])
plt.xlabel("Age Group")
plt.ylabel("Prefrence")
plt.title("Tea Preference")
plt.show()
```
